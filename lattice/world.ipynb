```python
# chunk overlay — verde=loaded, rojo=unloaded
world3 = World(64, 64, chunk_size=16)
world3.fill_rectangle(0, 0, 63, 63, 1)
world3.unload_chunk(2, 2)
world3.unload_chunk(3, 1)
world3.show(show_chunks=True)
```

```python
# force_load — cargar chunk y llenar
world4 = World(32, 32, chunk_size=16)
world4.unload_chunk(1, 1)

try:
    world4.fill_rectangle(16, 16, 31, 31, 2)
except ValueError as e:
    print(f"Error esperado: {e}")

world4.force_load(1, 1)
world4.fill_rectangle(16, 16, 31, 31, 2)
world4.show(show_chunks=True)
```
